```python
import matplotlib.pyplot as plt
from keras.preprocessing import image
from keras.preprocessing.image import ImageDataGenerator
%matplotlib inline

# EXERCISE:  Output four images rotated randomly between 0 and 360 degrees  =============
#  Load the image and change the size
img = image.load_img('./img/inu.png') 

# Convert to NumPy array, the shape of (height, width 3)
X = image.img_to_array(img) 

# Deform to (1, height, width, 3)
X = X.reshape((1,) + X.shape) 

# Instantiate IDG
datagen = ImageDataGenerator(rotation_range=360)

# Create the image of batch converted randomly
# We need to break around somewhere due to infinite loop 
i= 0
for batch in datagen.flow(X, batch_size=1): 
    plt.figure(i)
    imgplot = plt.imshow(image.array_to_img(batch[0]))
    i += 1
    if i == 4:
        break
plt.show()
# ==========================================
```

```python
# EXERCISE: 以下の様なディレクトリ構想になる様に画像を仕分けましょう。=======================
"""
/3rd_week/
                | -- dataset/
                                | -- flowers/
                                                | -- test_images/
                                                |                        | -- Bluebell/   <= 花の種別の名前(名前は `flower_labels.txtより`)
                                                |                        |                 | -- 
                                                |                        |                 | -- ............
                                                |                        | -- ClassName....
                                                | ....
                                                | -- train_images/
                                                |                        | -- Bluebell/
                                                ...................

"""

import os
import shutil
import random

IN_DIR = './dataset/flowers_origin'
TRAIN_DIR = './dataset/flowers/train_images'
TEST_DIR = './dataset/flowers/test_images'

if not os.path.exists('./dataset/flowers'):
    os.mkdir('./dataset/flowers')

if not os.path.exists(TRAIN_DIR):
    os.mkdir(TRAIN_DIR)

if not os.path.exists(TEST_DIR):
    os.mkdir(TEST_DIR)

# name => (start idx, end idx)
flower_dics = {}

with open('./static/flower_labels.txt') as fp:
    for line in fp:
        line = line.rstrip()
        cols = line.split()

        assert len(cols) == 3

        start = int(cols[0])
        end = int(cols[1])
        name = cols[2]

        flower_dics[name] = (start, end)

# # 花ごとのディレクトリを作成
# for name in flower_dics:
#     os.mkdir(os.path.join(TRAIN_DIR, name))
#     os.mkdir(os.path.join(TEST_DIR, name))

# データセットをスキャン
for f in sorted(os.listdir(IN_DIR)):
    # image_0001.jpg => 1
    prefix = f.replace('.jpg', ' ')
    idx = int(prefix.split('_')[1])

    for name in flower_dics:
        start, end = flower_dics[name]
        if idx in range(start, end + 1):
            source = os.path.join(IN_DIR, f)
            dest = os.path.join(TRAIN_DIR, name)
            shutil.copy(source, dest)
            continue

# 訓練データの各ディレクトリからランダムに10枚をテストとする
for d in os.listdir(TRAIN_DIR):
    files = os.listdir(os.path.join(TRAIN_DIR, d))
    random.shuffle(files)
    for f in files[:10]:
        source = os.path.join(TRAIN_DIR, d, f)
        dest = os.path.join(TEST_DIR, d)
        shutil.move(source, dest)
```

```python
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Activation, Dropout, Dense, Flatten

model = Sequential()

# EXERCISE: 上図の様なモデルを作りましょう ========================--
model.add(Conv2D(32, (3, 3), input_shape=(150, 150, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(32, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(64, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Flatten())
model.add(Dense(64))
model.add(Activation('relu'))
model.add(Dropout(0.5))
model.add(Dense(17))
model.add(Activation('softmax'))
# =======================================================
```

```python
# EXERCISE: 損失関数categorical_crossentropy, optimizerはAdamで多クラス分類用の学習プロセスを組んでコンパイルしましょう =====
model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
# ==========================================================================================
```

```python
from keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1.0 / 255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True)

test_datagen = ImageDataGenerator(rescale=1.0 / 255)
```

```python
classes_ = ['Tulip', 'Snowdrop', 'LilyValley', 'Bluebell', 'Crocus',
           'Iris', 'Tigerlily', 'Daffodil', 'Fritillary', 'Sunflower',
           'Daisy', 'ColtsFoot', 'Dandelion', 'Cowslip', 'Buttercup',
           'Windflower', 'Pansy']

train_generator = train_datagen.flow_from_directory(
    directory='./dataset/flowers/train_images',
    target_size=(150, 150),
    color_mode='rgb',
    classes=classes_,
    class_mode='categorical',
    batch_size=20,
    shuffle=True)

test_generator = test_datagen.flow_from_directory(
    directory='./dataset/flowers/test_images',
    target_size=(150, 150),
    color_mode='rgb',
    classes=classes_,
    class_mode='categorical',
    batch_size=20,
    shuffle=True)
```

```python
num_samples = len(images_) # 1360
num_trains = int(num_samples / 8 * 7)
num_vals = num_samples - num_trains
print(num_trains)
print(num_vals)
```

```python
history = model.fit_generator(train_generator,
                            steps_per_epoch=int(num_trains/ 20),
                            epochs=50,
                            validation_data=test_generator,
                            validation_steps=num_vals)
```

```python
from keras.layers import Input
from keras.models import Model
from keras.applications import VGG16
from keras import optimizers

# VGG16モデルと学習済み重みをロード
# Fully-connected層（FC）はいらないのでinclude_top=False）
input_tensor = Input(shape=(150, 150, 3))
vgg16 = VGG16(include_top=False, weights='imagenet', input_tensor=input_tensor)

# 全結合層を構築
top_model = Sequential()
top_model.add(Flatten(input_shape=vgg16.output_shape[1:]))
top_model.add(Dense(256, activation='relu'))
top_model.add(Dropout(0.5))
top_model.add(Dense(17, activation='softmax'))

# VGG16とFCを接続
model = Model(input=vgg16.input, output=top_model(vgg16.output))

# 最後のconv層の直前までの層をfreeze
for layer in model.layers[:-4]:
    layer.trainable = False

# Fine-tuningのときはSGDの方がよい
model.compile(loss='categorical_crossentropy',
              optimizer=optimizers.SGD(lr=1e-4, momentum=0.9),
              metrics=['accuracy'])
```

```python
# EXERCISE: KerasからVGG16をインポートして　 `base_model` としてインスタンス化しましょう ==============
from keras.applications import VGG16

base_model = VGG16(weights='imagenet',
                                      include_top=False,
                                      input_shape=(150, 150, 3))
#==========================================================================

# EXERCISE: 最後の４層以外を凍結しましょう ====================
base_model.trainable = True

for layer in base_model.layers[:-4]:
    layer.trainable = False
#=================================================
```

```python
from keras.models import Sequential

# EXERCISE: モデルを構築 ================
model = Sequential()

model.add(base_model)
model.add(Flatten())
model.add(Dense(256, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(17, activation='softmax'))
#=================================

model.summary()
```

```python
from keras.optimizers import SGD

model.compile(loss='categorical_crossentropy',
                          optimizer=SGD(lr=1e-4, momentum=0.9),
                          metrics=['accuracy'])
```
